tables or their columns, delete the model in the workspace first.

The model reads the Delta tables directly from OneLake (Direct Lake on OneLake). It needs the library
semantic-link-labs, which the first cell installs for this session.

```python
import sempy.fabric as fabric
import sempy_labs as labs
from sempy_labs import directlake
from sempy_labs.tom import connect_semantic_model

LAKEHOUSE = "WindFarmLH"            # the Lakehouse with the tables of notebook 05, in this workspace
MODEL = SPEC["name"]

missing = [t["lakehouseTable"] for t in SPEC["tables"] if not spark.catalog.tableExists(t["lakehouseTable"])]
assert not missing, f"Run notebook 05 first. Missing tables: {missing}"

# Model table name -> Lakehouse table. The library stops when a model of this name exists, so an
# existing model is kept, and the next cell replaces its relationships and measures.
EXISTS = MODEL in set(fabric.list_datasets(mode="rest")["Dataset Name"])
if EXISTS:
    print("Model exists:", MODEL, "- keeping its tables. To build them again, delete the model in the workspace first.")
else:
    directlake.generate_direct_lake_semantic_model(
        dataset=MODEL, tables={t["name"]: t["lakehouseTable"] for t in SPEC["tables"]}, source=LAKEHOUSE,
        source_type="Lakehouse", use_sql_endpoint=False, refresh=False)
    print("Model created:", MODEL)
```

## Relationships, measures

```python
with connect_semantic_model(dataset=MODEL, readonly=False) as tom:
    have = {t.Name for t in tom.model.Tables}
    want = {t["name"] for t in SPEC["tables"]}
    assert want <= have, f"The model lacks the tables {sorted(want - have)}. Delete the model in the workspace and run this notebook again."
    for name in [r.Name for r in tom.model.Relationships]:    # start clean, so that the notebook can run again
        tom.model.Relationships.Remove(name)
    for table in tom.model.Tables:
        for name in [m.Name for m in table.Measures]:
            table.Measures.Remove(name)

with connect_semantic_model(dataset=MODEL, readonly=False) as tom:
    for r in SPEC["relationships"]:
        tom.add_relationship(from_table=r["fromTable"], from_column=r["fromColumn"], to_table=r["toTable"],
                             to_column=r["toColumn"], from_cardinality="Many", to_cardinality="One",
                             is_active=r["active"])
    tom.mark_as_date_table(table_name="Date", column_name="Date")
    for table, column in SPEC["hidden"]:
        tom.model.Tables[table].Columns[column].IsHidden = True
    for table, column, by in SPEC["sortBy"]:
        tom.model.Tables[table].Columns[column].SortByColumn = tom.model.Tables[table].Columns[by]
    for m in SPEC["measures"]:
        tom.add_measure(table_name="KPIs", measure_name=m["name"], expression=m["expression"],
                        format_string=m["format"], description=m["description"], display_folder=m["folder"])
print(len(SPEC["relationships"]), "relationships and", len(SPEC["measures"]), "measures added")
```
